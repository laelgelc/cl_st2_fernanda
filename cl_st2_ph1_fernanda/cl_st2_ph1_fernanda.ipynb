## Build ENADE source file index

```python
from pathlib import Path

import pandas as pd

enade_raw_dir = Path("corpus/00_sources/enade_raw")

enade_df = pd.DataFrame(
    [
        {
            "year": file_path.parent.name,
            "filename_ori": file_path.name,
        }
        for file_path in enade_raw_dir.rglob("*")
        if file_path.is_file()
    ]
)

enade_df = enade_df.sort_values(["year", "filename_ori"]).reset_index(drop=True)

enade_df
```
