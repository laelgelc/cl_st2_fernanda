## Enrich ENADE metadata with mnemonics, disciplines, and standardized filenames

```python
# Enrich ENADE metadata with mnemonics, disciplines, and standardized filenames

from pathlib import Path

# Map mnemonic -> discipline, since enade_disciplines is {discipline: mnemonic}
mnemonic_to_discipline = {mnemonic: discipline for discipline, mnemonic in enade_disciplines.items()}

# Remove columns first if re-running this cell
enade_df = enade_df.drop(
    columns=[col for col in ["mnemonic", "discipline", "filename"] if col in enade_df.columns]
)

filename_ori_idx = enade_df.columns.get_loc("filename_ori")

# Create mnemonic column to the right of filename_ori
enade_df.insert(
    filename_ori_idx + 1,
    "mnemonic",
    enade_df["filename_ori"].map(enade_filenames)
)

# Create discipline column to the right of mnemonic
enade_df.insert(
    filename_ori_idx + 2,
    "discipline",
    enade_df["mnemonic"].map(mnemonic_to_discipline)
)

# Create standardized filename column to the right of discipline
years = enade_df["filename_ori"].str.extract(r"(\d{4})", expand=False)
extensions = enade_df["filename_ori"].apply(lambda filename: Path(filename).suffix.lstrip(".").lower())

enade_df.insert(
    filename_ori_idx + 3,
    "filename",
    "enade_" + years + "_" + enade_df["mnemonic"] + "." + extensions
)

enade_df
```

## Convert ENADE PDFs to text files

```python
# Convert ENADE PDFs to text files

from pathlib import Path

import pymupdf

enade_raw_dir = Path("corpus/00_sources/enade_raw")
enade_txt_dir = Path("corpus/01_enade_txt")

conversion_records = []

for row in enade_df.itertuples(index=False):
    year = str(row.year)
    source_pdf = enade_raw_dir / year / row.filename_ori
    output_dir = enade_txt_dir / year
    output_txt = output_dir / f"{Path(row.filename).stem}.txt"

    output_dir.mkdir(parents=True, exist_ok=True)

    if not source_pdf.exists():
        conversion_records.append({
            "year": year,
            "filename_ori": row.filename_ori,
            "filename": row.filename,
            "output_txt": str(output_txt),
            "status": "source_pdf_not_found",
        })
        continue

    try:
        with pymupdf.open(source_pdf) as pdf:
            text = "\n\n".join(page.get_text("text") for page in pdf)

        output_txt.write_text(text, encoding="utf-8")

        conversion_records.append({
            "year": year,
            "filename_ori": row.filename_ori,
            "filename": row.filename,
            "output_txt": str(output_txt),
            "status": "converted",
        })

    except Exception as error:
        conversion_records.append({
            "year": year,
            "filename_ori": row.filename_ori,
            "filename": row.filename,
            "output_txt": str(output_txt),
            "status": f"error: {error}",
        })

enade_conversion_df = pd.DataFrame(conversion_records)

enade_conversion_df["status"].value_counts()
```
