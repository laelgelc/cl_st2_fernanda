## Enrich ENADE metadata with mnemonics, disciplines, and standardized filenames

```python
# Enrich ENADE metadata with mnemonics, disciplines, and standardized filenames

from pathlib import Path

# Map mnemonic -> discipline, since enade_disciplines is {discipline: mnemonic}
mnemonic_to_discipline = {mnemonic: discipline for discipline, mnemonic in enade_disciplines.items()}

# Remove columns first if re-running this cell
enade_df = enade_df.drop(
    columns=[col for col in ["mnemonic", "discipline", "filename"] if col in enade_df.columns]
)

filepath_ori_idx = enade_df.columns.get_loc("filepath_ori")

# Create mnemonic column to the right of filepath_ori
enade_df.insert(
    filepath_ori_idx + 1,
    "mnemonic",
    enade_df["filename_ori"].map(enade_filenames)
)

# Create discipline column to the right of mnemonic
enade_df.insert(
    filepath_ori_idx + 2,
    "discipline",
    enade_df["mnemonic"].map(mnemonic_to_discipline)
)

# Create standardized filename column to the right of discipline
years = enade_df["filename_ori"].str.extract(r"(\d{4})", expand=False)
extensions = enade_df["filename_ori"].apply(lambda filename: Path(filename).suffix.lstrip(".").lower())

enade_df.insert(
    filepath_ori_idx + 3,
    "filename",
    "enade_" + years + "_" + enade_df["mnemonic"] + "." + extensions
)

enade_df
```

## Split ENADE PDFs into individual page PDFs

```python
from pathlib import Path

import fitz  # pymupdf
from tqdm.auto import tqdm


enade_pdf_pages_dir = Path("corpus/01_enade_pdf_pages")
enade_pdf_pages_dir.mkdir(parents=True, exist_ok=True)

page_records = []

for row in tqdm(enade_df.itertuples(index=False), total=len(enade_df), desc="Splitting ENADE PDFs"):
    source_pdf_path = Path(row.filepath_ori)
    output_filename = Path(row.filename)
    output_stem = output_filename.stem
    output_suffix = output_filename.suffix

    output_dir = enade_pdf_pages_dir / output_stem
    output_dir.mkdir(parents=True, exist_ok=True)

    with fitz.open(source_pdf_path) as source_doc:
        for page_idx in range(source_doc.page_count):
            page_number = page_idx + 1
            page_filename = f"{output_stem}_page_{page_number}{output_suffix}"
            page_filepath = output_dir / page_filename

            with fitz.open() as page_doc:
                page_doc.insert_pdf(source_doc, from_page=page_idx, to_page=page_idx)
                page_doc.save(page_filepath)

            page_records.append(
                {
                    "filename": row.filename,
                    "filepath_ori": row.filepath_ori,
                    "page_number": page_number,
                    "page_filename": page_filename,
                    "page_filepath": str(page_filepath),
                }
            )

enade_df_pdf_pages = pd.DataFrame(page_records)

print(f"Created {len(enade_df_pdf_pages)} page PDF files in {enade_pdf_pages_dir}")
enade_df_pdf_pages.head()
```
